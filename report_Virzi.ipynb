## **2.YSO science**

# **Acctretors**
Young Stellar Objects (YSOs) are stars in their early evolutionary stages, characterized by strong variability driven by circumstellar accretion, magnetic activity, and interactions with surrounding disk material. Spectroscopic indicators—particularly the $\text{H}\alpha$ emission line at $6562.8\ \text{\AA}$—play a fundamental role in tracing accretion processes. Broad $\text{H}\alpha$ profiles typically indicate high-velocity gas associated with magnetospheric accretion or outflows, whereas narrow profiles are generally linked to chromospheric activity, non-accreting stars (Reipurth et al. 1996), or residual nebular emission.

# **Diagnostic tools**

To identify in an accurate way active accretors, we measure the Full Width at Zero Intensity (FWZI, $\Delta\lambda$), which quantifies the total spectral extent of the emission line from the point where it first rises above the local continuum to where it returns to baseline. Because the line wings trace the highest-velocity gas in the magnetospheric accretion columns and stellar outflows, the FWZI directly probes the maximum kinematics of the line-forming region. This makes it particularly effective for characterizing broad $\text{H}\alpha$ profiles and distinguishing genuine accreting YSOs from narrow, low-velocity chromospheric or nebular emission.

## **4.Methods**

The data I have are from two different fields of the Rosette (star forming region 21909 and 22511). For each target, the FITS files provide two data products: the raw NOSS (No Sky Subtracted) spectra, which have no background correction applied at all (containing the star, the local nebula, and the Earth's atmosphere), and the spectra reduced with pipeline version 0.94+, which removes only the terrestrial atmospheric contribution while leaving the nebular emission intact. Although each pointing contains a large number of fibers covering various object types, my analysis only requires those classified as SKY, NEBULA, and YSO (whose exact selection criteria for the YSO flag are yet to be clarified). Having a clear overview of the spatial distribution of these three fiber classes across the field is essential to determine the optimal number and selection of nearby sky/nebular fibers to model and subtract the local background for each YSO.





In these distributions in Fig1 and Fig 2 we can see three types of points 
-**SKY**:fibers pointed at apparently empty regions of the sky to measure the Earth's atmospheric background emission  
-**NEBULA**:fibers deliberately placed on the ionized gas filaments and bright structures of the Rosette Nebula to characterize the local physical conditions and kinematics of the nebular emission itself. Together with the SKY fibers, they are used to estimate and subtract the local nebular background around each star
-**STAR YSO** :fibers targeting candidate young stars in the star-forming region  that are still surrounded by circumstellar accretion disks or envelopes, which constitute the primary science sample of the analysis

To ensure the highest accuracy possible when comparing each YSO candidate with its local background, I selected all SKY and NEBULA fibers located within a $0.25^\circ$ radius around the target. If no background fibers are found within this search radius, the candidate is directly discarded. Conversely, if one or more fibers are present, their spectra are averaged to construct a local mean background spectrum, which is then used to evaluate and compare the $\text{H}\alpha$ emission contribution.
